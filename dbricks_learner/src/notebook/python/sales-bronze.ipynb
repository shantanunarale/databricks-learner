```python
catalog = dbutils.widgets.get("catalog")
stream = dbutils.widgets.get("file_to_process")
```

```python
from pyspark.sql.functions import col, lit, expr, current_timestamp
```

```python
input_df = spark.read \
    .format("csv") \
        .options(**{
                "header": True,
                "escape": '"',
                "quote": '"'
                }) \
            .load(f"/Volumes/{catalog}/landing/retail_raw/{stream}.csv")
```

```python
input_df = input_df.withColumn("insert_time", current_timestamp())
```

```python
input_df.writeTo(f"{catalog}.bronze.{stream}").createOrReplace()
```
