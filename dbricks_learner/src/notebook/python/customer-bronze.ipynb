```python
from pyspark.sql.functions import col, lit, expr, current_timestamp
```

```python
catalog = dbutils.widgets.get("catalog")
```

```python
customer_df = spark.read \
    .format("csv") \
        .option("header", True) \
            .load(f"/Volumes/{catalog}/landing/retail_raw/customers.csv")
```

```python
customer_df = customer_df.withColumn("insert_time", current_timestamp())
```

```python
customer_df.write \
    .mode("overwrite") \
        .format("delta") \
            .saveAsTable(f"{catalog}.bronze.customers")
```
